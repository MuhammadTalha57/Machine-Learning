Implements Linear Regression algorithm for one variable using Gradient Descent.

```python
import math, copy
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Set
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])
```

```python
# Function to calculate cost
def compute_cost(x, y, w, b):
    m = x.shape[0]
    cost = 0
    for i in range(m):
        tmp_y = w * x[i] + b
        cost += (tmp_y - y[i])**2
    total_cost = cost / (2 * m)
    return total_cost
```

```python
# Function to calculate gradient
def compute_gradient(x, y, w, b):
    """
    Args:
        x (ndarray (m, )):  Data, m examples
        y (ndarray (m, )):  Target values
        w, b (scalar)    :  Model Parameters
    Returns
        dj_dw (scalar): The gradient of the cost w.r.t. the parameter w
        dj_db (scalar): The gradient of the cost w.r.t. the parameter b
    """

    m = x.shape[0]
    dj_dw = 0
    dj_db = 0
    for i in range(m):
        f_wb = w * x[i] + b
        dj_dw += (f_wb - y[i]) * x[i]
        dj_db += f_wb - y[i]
    
    dj_dw /= m
    dj_db /= m
    return dj_dw, dj_db
```

```python
def gradient_descent(x, y, w_in, b_in, alpha, num_iters, cost_function, gradient_function):
    b = b_in
    w = w_in

    for i in range(num_iters):
        dj_dw, dj_db = gradient_function(x, y, w, b)

        b = b - alpha * dj_db
        w = w - alpha * dj_dw

    return w, b
```

```python
# Initialize Params
w_init = 0.0
b_init = 0.0
iterations = 10000
tmp_alpha = 1.0e-2

w_final, b_final = gradient_descent(x_train, y_train, w_init, b_init, tmp_alpha, iterations, compute_cost, compute_gradient)

print(f"(w, b) found by gradient descent: ({w_final:8.4f},{b_final:8.4f})")
```
